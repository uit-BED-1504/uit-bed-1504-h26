## Oppgave 2.1 – Kodeforståelse

Hva blir verdien av `resultat` etter at koden er kjørt?

````python
profitt_marginer = [0.1, -0.05, 0.2, 0.05, -0.1]
resultat = 1
for margin in profitt_marginer:
    if margin > 0:
        resultat = resultat * (1 + margin)
print(resultat)
````

**Svar: `resultat` = 1,386.** Python skriver `1.3860000000000001` på grunn av flyttallsavrunding.

| `margin` | `> 0`? | `resultat` |
|---|---|---|
| 0.10 | ja | 1 · 1.10 = 1.10 |
| -0.05 | nei | 1.10 (uendret) |
| 0.20 | ja | 1.10 · 1.20 = 1.32 |
| 0.05 | ja | 1.32 · 1.05 = 1.386 |
| -0.10 | nei | 1.386 (uendret) |

Løkken går gjennom alle marginene, men `if`-testen tar bare med de positive. Hver positiv margin ganges inn som vekstfaktor $(1+m)$, så resultatet er en akkumulert vekst.

````python
# Sporing: skriv ut hvert steg
resultat = 1
for margin in profitt_marginer:
    if margin > 0:
        resultat = resultat * (1 + margin)
    print(f"margin = {margin:5.2f} -> resultat = {resultat:.4f}")
````

## Oppgave 2.2 – Pseudokode: total lønnskostnad

Lag pseudokode som regner ut den totale månedlige lønnskostnaden: timelønn × 150 timer for hver ansatt. Algoritmen skal skrive «Budsjett overskredet» hvis totalen er over 1 000 000 kr, og ellers skrive ut totalen.

```
INNDATA: timelonner   // timelønn per ansatt
SETT timer         TIL 150
SETT budsjett      TIL 1 000 000
SETT total_kostnad TIL 0

FOR HVER lonn I timelonner:
    SETT maanedslonn   TIL lonn * timer
    SETT total_kostnad TIL total_kostnad + maanedslonn
SLUTT FOR

HVIS total_kostnad > budsjett:
    SKRIV "Budsjett overskredet"
ELLERS:
    SKRIV total_kostnad
SLUTT HVIS
```

**Typisk feil:** Å sjekke budsjettet *inne i* løkken. Da kan meldingen skrives ut flere ganger, og før totalen er ferdig summert.

Samme algoritme i Python:

````python
def total_lonnskostnad(timelonner, timer=150, budsjett=1_000_000):
    total = 0
    for lonn in timelonner:
        total += lonn * timer
    if total > budsjett:
        print("Budsjett overskredet")
    else:
        print(f"Total kostnad: {total} kr")
    return total

total_lonnskostnad([250, 300, 420])
total_lonnskostnad([500] * 20)   # 20 ansatte à 500 kr/t -> over budsjett
````

## Oppgave 2.4 – Feilsøking

Den opprinnelige koden har to feil:

```python
def beregn_naaverdi(sluttverdi, rente, aar)
naaverdi = sluttverdi / (1 + rente)**aar
    return naaverdi
```

1. **Kolon mangler** på slutten av `def`-linjen. Det gir `SyntaxError`.
2. **Feil innrykk:** linje 2 er ikke rykket inn og hører derfor ikke til funksjonskroppen (`IndentationError`). Linje 2 og 3 må ha samme innrykk.

Formelen er riktig: nåverdi = $K_T / (1+r)^T$.

**Rettet kode:**

````python
def beregn_naaverdi(sluttverdi, rente, aar):
    naaverdi = sluttverdi / (1 + rente)**aar
    return naaverdi

print(beregn_naaverdi(1000, 0.05, 10))
````

````python
# Demonstrasjon: den opprinnelige koden gir en feil
feil_kode = '''def beregn_naaverdi(sluttverdi, rente, aar)
naaverdi = sluttverdi / (1 + rente)**aar
    return naaverdi
'''
try:
    exec(feil_kode)
except SyntaxError as e:
    print(type(e).__name__, "-", e.msg, "(linje", e.lineno, ")")
````

## Oppgave 2.5 – Marginalkostnad med sympy

Kostnadsfunksjon: $C(Q) = 100 + 10Q + 0{,}5Q^2$. Marginalkostnaden er den deriverte, $MC(Q) = C'(Q)$.

````python
import sympy as sp

Q = sp.symbols('Q')            # steg 1: definer symbolet
C = 100 + 10*Q + 0.5*Q**2      # steg 2: kostnadsfunksjonen
MC = sp.diff(C, Q)             # steg 3: deriver
print(MC)                      # steg 4: MC(Q)
print(MC.subs(Q, 20))          # steg 5: MC i Q = 20
````

**Resultat:** $MC(Q) = 10 + Q$. Sympy skriver `1.0*Q + 10` fordi `0.5` er et desimaltall.

**Tolkning:** Ved $Q = 20$ koster det ca. 30 kr ekstra å produsere én enhet til.

````python
# Bonus: tegn C(Q) og MC(Q)
C_f  = sp.lambdify(Q, C)
MC_f = sp.lambdify(Q, MC)
q = np.linspace(0, 50, 200)

fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].plot(q, C_f(q)); ax[0].set_title("Kostnad C(Q)"); ax[0].set_xlabel("Q")
ax[1].plot(q, MC_f(q), color="tab:green"); ax[1].set_title("Marginalkostnad MC(Q)"); ax[1].set_xlabel("Q")
ax[1].scatter([20], [MC_f(20)], color="red", zorder=3)
ax[1].annotate("MC(20) = 30", (20, 30), xytext=(25, 25))
plt.tight_layout()
plt.show()
````

## Oppgave 3.1 – Nedbetaling av lån

**Krav:**
- Funksjonen `beregn_nedbetalingstid(laanebelop, aarlig_rente, maanedsbelop)`
- En `while`-løkke med én runde per måned
- Renten legges til **før** innbetalingen trekkes fra
- Månedlig rente = `aarlig_rente / 12`
- Funksjonen returnerer antall måneder

**Algoritme for én måned:** saldo ← saldo · (1 + r_m) − månedsbeløp. Gjenta så lenge saldoen er over 0.

**Fallgruve:** Hvis månedsbeløpet er mindre enn eller lik renten for første måned, blir gjelden aldri mindre, og løkken går evig. Derfor sjekker vi det først.

````python
def beregn_nedbetalingstid(laanebelop, aarlig_rente, maanedsbelop):
    """Returnerer antall måneder det tar å nedbetale et lån."""
    maanedlig_rente = aarlig_rente / 12        # konverter årlig rente til månedlig
    saldo = laanebelop                         # gjenstående gjeld
    antall_maaneder = 0                        # teller

    # Sikkerhetssjekk: innbetalingen må være større enn første måneds rente,
    # ellers blir gjelden aldri mindre og løkken går evig.
    if maanedsbelop <= saldo * maanedlig_rente:
        raise ValueError("Månedsbeløpet er for lavt - lånet blir aldri nedbetalt.")

    while saldo > 0:                           # så lenge det er gjeld igjen
        saldo = saldo + saldo * maanedlig_rente    # 1) legg til månedens rente
        saldo = saldo - maanedsbelop               # 2) trekk fra innbetalingen
        antall_maaneder += 1                       # 3) én måned har gått

    return antall_maaneder
````

````python
print(beregn_nedbetalingstid(100_000, 0.05, 2_000))     # eksempelet i oppgaven
print(beregn_nedbetalingstid(10_000, 0.0, 1_000))       # kontroll: 0 % rente -> 10 måneder
print(beregn_nedbetalingstid(2_000_000, 0.05, 15_000))  # boliglån-eksempel
````

Et lån på 100 000 kr med 5 % rente og 2 000 kr i måneden er nedbetalt etter **57 måneder** (≈ 4,75 år). Kontrollen med 0 % rente gir nøyaktig 10 måneder, som forventet.

**Vanlige feil:** å bruke årlig rente direkte (glemme `/12`), å trekke fra innbetalingen før renten legges til, å glemme å øke telleren, og å bruke `print` i stedet for `return`.

````python
# Sikkerhetssjekken i praksis: for lavt månedsbeløp
try:
    beregn_nedbetalingstid(100_000, 0.05, 400)   # renten første måned er ca. 417 kr
except ValueError as e:
    print("Feil:", e)
````

````python
# Bonus: hvordan saldoen utvikler seg måned for måned
def saldoforlop(laanebelop, aarlig_rente, maanedsbelop):
    r = aarlig_rente / 12
    saldo, forlop = laanebelop, [laanebelop]
    while saldo > 0:
        saldo = saldo * (1 + r) - maanedsbelop
        forlop.append(max(saldo, 0))
    return forlop

forlop = saldoforlop(100_000, 0.05, 2_000)
plt.plot(forlop)
plt.title("Saldo for lån på 100 000 kr, 5 % rente, 2 000 kr/mnd")
plt.xlabel("Måned"); plt.ylabel("Gjenstående saldo (kr)")
plt.show()
````

## Oppgave 3.2 – Dataanalyse med pandas

a) Lag en ny kolonne `Profitt` = `Inntekter` − `Kostnader`.
b) Regn ut total profitt per produktkategori.
c) Finn kategorien med høyest total profitt.

````python
import pandas as pd
data = {
    'Dato': ['2024-05-01', '2024-05-01', '2024-05-02', '2024-05-03', '2024-05-03'],
    'Produktkategori': ['Elektronikk', 'Matvarer', 'Elektronikk', 'Klær', 'Matvarer'],
    'Inntekter': [5000, 1500, 8000, 3000, 2000],
    'Kostnader': [3500, 1200, 6000, 1800, 1600]
}
salgsdata = pd.DataFrame(data)
salgsdata
````

````python
# a) Ny kolonne: Profitt = Inntekter - Kostnader (radvis, uten løkke)
salgsdata['Profitt'] = salgsdata['Inntekter'] - salgsdata['Kostnader']
salgsdata
````

````python
# b) Total profitt per produktkategori
profitt_per_kategori = salgsdata.groupby('Produktkategori')['Profitt'].sum()
print(profitt_per_kategori)
````

````python
# c) Kategorien med høyest total profitt
beste_kategori = profitt_per_kategori.idxmax()
print(f"Kategorien med høyest profitt er: {beste_kategori}")
````

**Nøkkelideer:**
- **a)** Kolonnene trekkes fra hverandre element for element, så vi trenger ingen løkke.
- **b)** `groupby` + `sum` gir Elektronikk 1500 + 2000 = **3500**, Matvarer 300 + 400 = **700** og Klær **1200**.
- **c)** `idxmax()` gir *indeksen*, altså kategorinavnet. `max()` gir bare tallet 3500.

Alternativ løsning for c): `profitt_per_kategori.sort_values(ascending=False).index[0]`

````python
# Bonus: visualisering av profitt per kategori
profitt_per_kategori.sort_values().plot(kind='barh', color='tab:green')
plt.title("Total profitt per produktkategori")
plt.xlabel("Profitt (kr)")
plt.show()
````
