## Oppgave 2.2 – Simulering (2 p)

**Hva er simulering?** Vi etterligner et fenomen fra virkeligheten med en datamodell, ofte med *tilfeldighet* (`np.random`) og *utvikling over tid*. Vi gjentar prosessen mange ganger og ser på fordelingen av utfallene.

**Hvorfor?** Mange problemer har ingen enkel analytisk løsning, eller de er for dyre eller umulige å teste i virkeligheten.

**Hvorfor matplotlib?** Tusenvis av tall er vanskelige å tolke. Plott viser forløp, spredning og usikkerhet.

**To eksempler** (kjør cellene under):
1. Investering med tilfeldig avkastning over 10 år
2. Terningkast og loven om store tall

**Sensor:** Full pott for gjentatte prosesser og tilfeldighet, plotting, og *hvorfor*.

````python
# Eksempel 1: 1000 simulerte investeringer på 10 000 kr over 10 år
np.random.seed(1)
n_sim, aar = 1000, 10
avkastning = np.random.normal(loc=0.06, scale=0.15, size=(n_sim, aar))   # tilfeldig årlig avkastning
verdi = 10_000 * np.cumprod(1 + avkastning, axis=1)

fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
ax[0].plot(range(1, aar + 1), verdi[:50].T, alpha=0.4)
ax[0].set_title("50 simulerte baner"); ax[0].set_xlabel("År"); ax[0].set_ylabel("Verdi (kr)")
ax[1].hist(verdi[:, -1], bins=40)
ax[1].set_title("Sluttverdi etter 10 år"); ax[1].set_xlabel("Verdi (kr)")
plt.tight_layout(); plt.show()
print(f"Median sluttverdi: {np.median(verdi[:, -1]):.0f} kr, "
      f"andel som taper penger: {(verdi[:, -1] < 10_000).mean():.1%}")
````

````python
# Eksempel 2: terningkast og loven om store tall
np.random.seed(2)
kast = np.random.randint(1, 7, size=5000)
snitt = np.cumsum(kast) / np.arange(1, kast.size + 1)
plt.plot(snitt, label="Løpende gjennomsnitt")
plt.axhline(3.5, color="red", ls="--", label="Forventning 3,5")
plt.xscale("log"); plt.xlabel("Antall kast"); plt.legend(); plt.title("Loven om store tall")
plt.show()
````

## Oppgave 2.5 – Feilsøking: frikort (4 p)

```python
def sjekk_frikort(lønn)
if lønn < 100000:
     resultat = Du får frikort
else:
     resultat = Du må betale skatt
     return resultat
```

**Fire feil (1 poeng hver):**
1. **Kolon mangler** etter `def`-linjen.
2. **Innrykk mangler:** `if`/`else` hører ikke til funksjonen.
3. **Anførselstegn mangler:** teksten må være en streng.
4. **`return` står feil:** den ligger inne i `else`, så lønn under 100 000 gir `None`.

````python
def sjekk_frikort(lønn):
    if lønn < 100000:
        resultat = "Du får frikort"
    else:
        resultat = "Du må betale skatt"
    return resultat

print(sjekk_frikort(80_000))
print(sjekk_frikort(450_000))
````

````python
# Hva skjer hvis bare feil 4 gjenstår (return inne i else)?
def sjekk_frikort_feil(lønn):
    if lønn < 100000:
        resultat = "Du får frikort"
    else:
        resultat = "Du må betale skatt"
        return resultat

print(sjekk_frikort_feil(80_000))    # None!
````

## Oppgave 2.6 – sympy: stasjonære punkter (3 p)

````python
import sympy as sp
x = sp.symbols('x')
f = x**3 - 3*x**2 + 2
df = sp.diff(f, x)
pnkt = sp.solve(sp.Eq(df, 0), x)
print(pnkt)
````

| Linje | Forklaring |
|---|---|
| 1 | Importerer SymPy som `sp`. |
| 2 | Gjør `x` til et matematisk symbol. |
| 3 | $f(x) = x^3 - 3x^2 + 2$ |
| 4 | Deriverer: $f'(x) = 3x^2 - 6x$. (`df` betyr her «derivert», ikke DataFrame.) |
| 5 | `sp.Eq(df, 0)` lager ligningen $f'(x)=0$, og `sp.solve` løser den. |
| 6 | Skriver ut `[0, 2]`. |

$3x^2 - 6x = 3x(x-2) = 0 \Rightarrow x = 0$ eller $x = 2$.

**Sensor (3 p):** Derivasjon, at den deriverte settes lik 0, og løsning for $x$.

````python
# Ekstra: klassifiser punktene med den andrederiverte og tegn funksjonen
f2 = sp.diff(f, x, 2)
for p in pnkt:
    type_ = "toppunkt" if f2.subs(x, p) < 0 else "bunnpunkt"
    print(f"x = {p}: f(x) = {f.subs(x, p)}, f''(x) = {f2.subs(x, p)} -> {type_}")

f_num = sp.lambdify(x, f)
xs = np.linspace(-1.2, 3.2, 200)
plt.plot(xs, f_num(xs))
plt.scatter([0, 2], [f_num(0), f_num(2)], color="red", zorder=3)
plt.axhline(0, color="grey", lw=0.5)
plt.title("f(x) = x³ − 3x² + 2 med stasjonære punkter"); plt.show()
````

## Oppgave 3.1 – Bankrådgiver

**Krav:** funksjonen `sparemål(antatt_rente, startkapital, sparemål)` med en `while`-løkke. Den skal returnere antall år, renten og startkapitalen, og resultatet skal skrives ut med `print`.

**Kontroll med formel:** $10\,000 \cdot 1{,}05^n \ge 20\,000 \iff n \ge \ln 2/\ln 1{,}05 \approx 14{,}2$, altså **15 år**.

````python
# Definerer funksjonen med de oppgitte parameterne
def sparemål(antatt_rente, startkapital, sparemål):
    kapital = startkapital     # kapital starter på startkapitalen
    år = 0                     # teller antall år

    # Så lenge vi ikke har nådd målet: legg til ett års rente
    while kapital < sparemål:
        kapital += kapital * antatt_rente   # rentes rente
        år += 1                             # ett år har gått

    # Returnerer antall år, renten og startkapitalen
    return år, antatt_rente, startkapital

# Kaller funksjonen og pakker ut de tre returverdiene
år, rente, start = sparemål(0.05, 10000, 20000)

print(f"Du når sparemålet på 20000 kr etter {år} år, "
      f"med startkapital på {start:.2f} kr og rente på {rente*100:.0f} %.")
````

````python
# Kontroll med formel
import math
print(math.log(2) / math.log(1.05), "->", math.ceil(math.log(2) / math.log(1.05)), "år")
````

**Merk:** Funksjonen og parameteren heter begge `sparemål`. Det virker, men er forvirrende. Et bedre navn er for eksempel `beregn_sparetid`.

**Sensor (0–5 p):** 5 p for forstått oppgave, gode kommentarer og riktig syntaks. Glemt kolon eller parentes gir lite trekk. Glemt innrykk, `return` eller parametere gir mer.
(Merk: sensorveiledningen skriver «maks 3 poeng», men skalaen går til 5, og Del 3 har maks 10 poeng totalt.)

````python
# Bonus: kapitalen år for år
k, forlop = 10_000, [10_000]
while k < 20_000:
    k *= 1.05
    forlop.append(k)
plt.plot(forlop, marker="o")
plt.axhline(20_000, color="red", ls="--", label="Sparemål")
plt.xlabel("År"); plt.ylabel("Kapital (kr)"); plt.legend()
plt.title("10 000 kr med 5 % rente"); plt.show()
````

## Oppgave 3.2 – Salgsdata

1. Lag en kolonne `Pris_EU` med pris i euro (1 € = 11,63 NOK).
2. Bruk en for-løkke til å lage en liste med varene som koster **mer enn** 700 kr.

````python
import pandas as pd

# (Kjørbar versjon: lager dataene i stedet for read_csv)
df = pd.DataFrame({
    'Vare': ['Jakke', 'Sko', 'Sekk', 'Lue', 'Caps'],
    'Pris': [1200, 800, 650, 250, 350],
    'Antall_varer_solgt': [5, 8, 4, 12, 16]})
df.head()
````

````python
# 1) Ny kolonne med pris i euro: DEL på kursen
valutakurs = 11.63
df['Pris_EU'] = round(df['Pris'] / valutakurs, 2)

# 2) For-løkke: finn varer som koster mer enn 700 kr
varer_over_700 = []                        # tom liste
for i in range(len(df)):                   # i = 0, 1, ..., 4
    if df.loc[i, 'Pris'] > 700:            # sjekk prisen i rad i
        varer_over_700.append(df.loc[i, 'Vare'])

print(df)
print("Varer som koster mer enn 700 kr:", varer_over_700)
````

- `df['Pris_EU'] = ...` lager en ny kolonne. Delingen skjer vektorisert, for alle rader samtidig.
- `range(len(df))` gir radindeksene, og `df.loc[i, 'Pris']` henter én celle.
- Sekk (650 kr) er ikke med, fordi grensen er *mer enn* 700.

**Sensor (0–5 p):** 5 p når logikken er på plass og studenten behersker `for` og `if`.

````python
# Alternativ uten løkke (oppgaven krevde løkke, men dette er godt å kunne)
print(df.loc[df['Pris'] > 700, 'Vare'].tolist())

# Bonus: omsetning per vare
df['Omsetning'] = df['Pris'] * df['Antall_varer_solgt']
df.plot(x='Vare', y='Omsetning', kind='bar', legend=False, title='Omsetning per vare (NOK)')
plt.show()
````

# Karaktersetting

$$\text{Total \%} = \frac{P_1}{40}\cdot 0{,}40 + \frac{P_2}{16}\cdot 0{,}30 + \frac{P_3}{10}\cdot 0{,}30$$

| Poengsum | Karakter |
|---|---|
| 92–100 % | A |
| 77–91 % | B |
| 58–76 % | C |
| 46–57 % | D |
| 40–45 % | E |
| 0–39 % | F |

**Merk:** Del 2 gir $2+2+3+3+4+3 = 17$ poeng, men formelen bruker 16.

````python
def karakter(p1, p2, p3, maks=(40, 16, 10), vekt=(0.40, 0.30, 0.30)):
    total = sum(p / m * v for p, m, v in zip((p1, p2, p3), maks, vekt)) * 100
    for grense, k in [(92, 'A'), (77, 'B'), (58, 'C'), (46, 'D'), (40, 'E')]:
        if total >= grense:
            return round(total, 1), k
    return round(total, 1), 'F'

print(karakter(30, 12, 6))     # eksempelet fra sensorveiledningen: 70.5 % -> C
print(karakter(38, 15, 9))
````
