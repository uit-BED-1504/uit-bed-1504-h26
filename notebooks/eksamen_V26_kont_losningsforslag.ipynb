## Oppgave 2.1 – Mutable og immutable (3 p)

- **Mutable** (foranderlig): kan endres etter at objektet er laget, for eksempel `list`, `dict`, `set` og `np.array`.
- **Immutable** (uforanderlig): kan ikke endres, og en «endring» lager et nytt objekt, for eksempel `int`, `float`, `str`, `tuple` og `bool`.

````python
# Mutable: liste
liste = [1, 2, 3]
liste[0] = 100          # lov, endres på stedet
print(liste)

# Immutable: streng
tekst = "Hei"
try:
    tekst[0] = "h"
except TypeError as e:
    print("TypeError:", e)

tekst = "h" + tekst[1:]  # lager en NY streng
print(tekst)
````

## Oppgave 2.2 – for- og while-løkker (6 p)

**Hensikt:** å gjenta en kodeblokk uten å skrive den om igjen.

- **for:** når antall runder er kjent, eller vi skal gå gjennom en sekvens.
- **while:** så lenge en betingelse er `True`, og antall runder er ukjent på forhånd.

**Typisk feil:** `while` uten oppdatering av variabelen gir en evig løkke.

````python
# for: kjent antall runder
for i in range(3):
    print(f"Runde nummer {i}")

# while: til betingelsen brytes
teller = 0
while teller < 3:
    print(f"Teller er {teller}")
    teller += 1     # unngår evig løkke
````

````python
# Typisk while-bruk: ukjent antall runder (hvor mange doblinger før vi passerer 1000?)
verdi, doblinger = 1, 0
while verdi < 1000:
    verdi *= 2
    doblinger += 1
print(doblinger, verdi)
````

## Oppgave 2.4 – Feilsøking: budsjett (6 p)

```python
def budsjett(lonn, utgift)
    penger_igjen = utgift - lonn
    return penger_igjen

lonn = 30000
utgift = 25000
print(f'Hvis jeg tjener {lonn} kr, og har {utgift} kr i utgift - da sitter jeg igjen med {penger_igjen}.'
```

**Fire feil:**
- (a) Kolon mangler etter `def`-linjen (syntaks).
- (b) `utgift - lonn` gir −5000. Det skal være `lonn - utgift` (logikk).
- (c) Avsluttende `)` mangler i `print` (syntaks).
- (d) `penger_igjen` er lokal i funksjonen, og funksjonen blir aldri kalt. Det gir `NameError` (scope).

````python
def budsjett(lonn, utgift):          # (a) la til kolon
    penger_igjen = lonn - utgift     # (b) lønn minus utgift
    return penger_igjen

lonn = 30000
utgift = 25000

resultat = budsjett(lonn, utgift)    # (d) kall funksjonen og lagre returverdien

print(f'Hvis jeg tjener {lonn} kr, og har {utgift} kr i utgift - '
      f'da sitter jeg igjen med {resultat}.')                    # (c) la til )
````

````python
# Demonstrasjon av scope-feilen (d): lokale variabler finnes ikke utenfor funksjonen
def demo():
    lokal = 123
    return lokal

demo()
try:
    print(lokal)
except NameError as e:
    print("NameError:", e)
````

## Oppgave 2.6 – if / elif / else i en løkke (6 p)

````python
tall_liste = [10, 5, 8, 12, 3]
resultat = 0

for x in tall_liste:
    if x > 8:
        resultat += x
    elif x % 2 != 0:
        resultat -= x
    else:
        resultat += 1

print(resultat)
````

````python
# Sporing steg for steg
resultat = 0
for x in tall_liste:
    if x > 8:
        resultat += x;  gren = f"if   (x > 8)   -> +{x}"
    elif x % 2 != 0:
        resultat -= x;  gren = f"elif (oddetall) -> -{x}"
    else:
        resultat += 1;  gren = "else           -> +1"
    print(f"x = {x:2d}: {gren:22s} resultat = {resultat}")
````

Bare den **første** sanne grenen kjøres. For `x = 8` er `8 > 8` usant og 8 er partall, så vi havner i `else`.

## Oppgave 3.1 – Sparing med årlig innskudd

**Krav:** funksjonen `sparing_med_innskudd(startkapital, arlig_innskudd, rente, sparemal)` med en `while`-løkke. **Renten legges til først**, deretter innskuddet. Funksjonen returnerer antall år.

Ett år i løkken: saldo ← saldo · (1 + r) + innskudd

````python
def sparing_med_innskudd(startkapital, arlig_innskudd, rente, sparemal):
    saldo = startkapital          # starter med startkapitalen
    ar = 0                        # teller antall år

    # Kjører så lenge saldoen er under sparemålet
    while saldo < sparemal:
        saldo = saldo * (1 + rente)      # 1) renter først
        saldo = saldo + arlig_innskudd   # 2) deretter innskudd
        ar += 1                          # 3) ett år har gått

    return ar                     # antall år (heltall)

aar_til_mal = sparing_med_innskudd(5000, 2000, 0.04, 30000)
print(f"Det tar {aar_til_mal} år å nå sparemålet på 30 000 kr.")
````

````python
# Saldo år for år
saldo, rader = 5000, [(0, 5000.0)]
while saldo < 30000:
    saldo = saldo * 1.04 + 2000
    rader.append((len(rader), round(saldo, 2)))
pd.DataFrame(rader, columns=["År", "Saldo (kr)"])
````

````python
# Betyr rekkefølgen noe?
def antall_ar(s0, inn, r, mal, rente_forst=True):
    s, a = s0, 0
    while s < mal:
        s = s * (1 + r) + inn if rente_forst else (s + inn) * (1 + r)
        a += 1
    return a, round(s, 2)

print("Rente først, så innskudd:", antall_ar(5000, 2000, 0.04, 30000))
print("Innskudd først, så rente:", antall_ar(5000, 2000, 0.04, 30000, rente_forst=False))
print("Ingen rente:             ", antall_ar(5000, 2000, 0.00, 30000))
````

Her gir begge rekkefølgene **10 år**, men saldoen blir forskjellig (31 413 kr mot 32 374 kr). Med andre tall kan også antall år bli forskjellig, så rekkefølgen oppgaven krever, må følges.

**Sensor ser etter:** en funksjon med fire parametere og `return`, en korrekt `while`-betingelse, rente før innskudd, en teller, et funksjonskall med `print`, og kommentarer.

````python
# Bonus: visualisering
saldoer = [r[1] for r in rader]
plt.plot(saldoer, marker="o", label="Saldo")
plt.axhline(30000, color="red", ls="--", label="Sparemål 30 000 kr")
plt.xlabel("År"); plt.ylabel("Kr"); plt.legend()
plt.title("5 000 kr + 2 000 kr/år, 4 % rente"); plt.show()
````

## Oppgave 3.2 – Analyse av eksamensresultater

1. Lag en ny kolonne `Bestått`: `True` hvis poeng ≥ 40.
2. Bruk en for-løkke til å lage en liste over studentene som ikke har bestått.
3. Skriv ut listen.

````python
import pandas as pd

# (Kjørbar versjon: lager dataene i stedet for read_csv)
df = pd.DataFrame({
    "Student": ["Ola", "Kari", "Per", "Lise", "Espen"],
    "Poeng":   [45, 82, 35, 39, 90]})

# 1) Ny kolonne: True hvis poeng >= 40, ellers False
df['Bestått'] = df['Poeng'] >= 40

# 2) For-løkke: finn studenter som IKKE bestod
ikke_bestatt_liste = []                  # tom liste
for i in range(len(df)):                 # i = 0, 1, ..., 4
    if df.loc[i, 'Poeng'] < 40:          # under grensen?
        navn = df.loc[i, 'Student']      # hent navnet
        ikke_bestatt_liste.append(navn)  # legg til i listen

# 3) Skriv ut listen
print(df)
print(f"Studenter som ikke bestod: {ikke_bestatt_liste}")
````

- `df['Poeng'] >= 40` gir en **boolsk** Series som blir den nye kolonnen (vektorisert).
- `range(len(df))` gir radindeksene, og `df.loc[i, 'Poeng']` henter én celle.
- Grensen er **40 eller mer**, så Lise (39) stryker.

````python
# Alternativ uten løkke (godt å kunne, men oppgaven krevde løkke)
print(df.loc[~df['Bestått'], 'Student'].tolist())

# Bonus: visualisering med beståttgrense
farger = ['tab:green' if b else 'tab:red' for b in df['Bestått']]
plt.bar(df['Student'], df['Poeng'], color=farger)
plt.axhline(40, color='black', ls='--', label='Grense 40 poeng')
plt.ylabel('Poeng'); plt.title('Eksamensresultater'); plt.legend(); plt.show()
````
