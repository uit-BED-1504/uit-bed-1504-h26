### Grenseverdier: `>` eller `>=`?
- «minst», «fra og med», «eller mer» → `>=`
- «over», «mer enn» → `>`

Test alltid **akkurat på grensen**.

````python
def gratis_frakt(belop):
    return belop >= 399      # "gratis frakt fra og med 399 kr"

for b in [398, 399, 400]:
    print(b, gratis_frakt(b))
````

### `return` i en løkke kan avslutte funksjonen med én gang

### Robuste funksjoner: riktig nevner og tomt utvalg
To klassiske feil i gjennomsnitt: å dele på lengden av *hele* listen i stedet for utvalget, og å glemme at utvalget kan være tomt.

````python
def snitt_over_feil(tall, grense):
    utvalg = []
    for t in tall:
        if t > grense:
            utvalg.append(t)
    return sum(utvalg) / len(tall)        # feil nevner!

def snitt_over(tall, grense):
    utvalg = []
    for t in tall:
        if t > grense:
            utvalg.append(t)
    if not utvalg:                        # tom liste er "falsy"
        return None
    return sum(utvalg) / len(utvalg)

print(snitt_over_feil([2, 10, 20], 5))
print(snitt_over([2, 10, 20], 5))
print(snitt_over([1, 2], 5))
````

## 7. Algoritmemønstre i løkker

Nesten alle løkkeoppgaver bygger på noen få mønstre. Oppskriften er:
1. Startverdi **før** løkken
2. Oppdatering **inne i** løkken
3. Bruk/returner **etter** løkken

### Akkumulator, teller og samleliste

````python
data = [4, 7, 2, 9]
total = 0
antall_store = 0
partall = []
for x in data:
    total += x
    if x > 5:
        antall_store += 1
    if x % 2 == 0:
        partall.append(x)
print(total, antall_store, partall)
````

### «Beste hittil» og indekser
Husk **indeksen** til det beste elementet. Bruk en «ikke funnet»-verdi (`-1` eller `None`) så tom liste håndteres.

````python
def varmeste_dag(temperaturer):
    beste = -1
    for i in range(len(temperaturer)):
        if beste == -1 or temperaturer[i] > temperaturer[beste]:
            beste = i
    return beste

print(varmeste_dag([12, 18, 15, 18]))
print(varmeste_dag([]))
````

**🧠 Forutsi først:** Hvorfor blir svaret 1 og ikke 3, når både indeks 1 og 3 har 18 grader? Hva skjer med `>=`?

<details><summary>Vis svar</summary>

Med `>` erstattes den beste bare av noe *strengt større*, så den første beholdes. Med `>=` ville indeks 3 vunnet.

</details>

`enumerate` gir indeks og verdi samtidig:

````python
for i, t in enumerate([12, 18, 15]):
    print(i, t)
````

### Tilstand som oppdateres og nullstilles
To variabler: én for *nåværende* tilstand og én for *beste* hittil. Still deg spørsmålene: Hva bryter tilstanden? Hva om den varer helt til slutt? Hva om listen er tom?

````python
def lengste_rekke_store(ordliste):
    nå = 0
    rekord = 0
    for ord in ordliste:
        if ord[0].isupper():
            nå += 1
            rekord = max(rekord, nå)
        else:
            nå = 0
    return rekord

print(lengste_rekke_store(["Oslo", "Bergen", "og", "Tromsø", "Bodø", "Alta", "by"]))
print(lengste_rekke_store([]))
````

### Tilfeldige tall og simulering
- `np.random.randint(lav, høy, size=n)`: `lav` er med, `høy` er **ikke** med.
- En andel over mange simuleringer er et **estimat** av en sannsynlighet.

````python
np.random.seed(42)
terning1 = np.random.randint(1, 7, size=100_000)
terning2 = np.random.randint(1, 7, size=100_000)
sum_to = terning1 + terning2
print("minste og største:", sum_to.min(), sum_to.max())
print("P(sum >= 10) ca.", (sum_to >= 10).mean(), " teoretisk:", 6/36)
````

### Hvor plasseres `seed`?
`seed` setter generatorens starttilstand. Settes den inne i løkken, nullstilles generatoren hver runde.

````python
# Seed inne i løkken: samme tall hver gang
kast = []
for i in range(5):
    np.random.seed(1)
    kast.append(np.random.randint(1, 7))
print("seed inne i løkken:", kast)

# Seed én gang før løkken: reproduserbart, men generatoren går videre
np.random.seed(1)
kast = []
for i in range(5):
    kast.append(np.random.randint(1, 7))
print("seed før løkken:   ", kast)
````

### Et lite økonomisk simuleringseksempel

````python
np.random.seed(0)
kunder = np.random.randint(20, 61, size=365)   # 20 til og med 60 kunder per dag
inntekt = 85 * kunder                          # 85 kr per kunde
resultat = inntekt - 3000                      # faste kostnader per dag
print("Andel dager med overskudd:", (resultat > 0).mean())
print("Gjennomsnittlig resultat:", resultat.mean())
````

## 10. SymPy

`import sympy as sp` **importerer** biblioteket (det må være installert fra før).

### Likninger
`sp.solve(uttrykk, x)` løser `uttrykk = 0` og returnerer alltid en **liste**.

````python
x = sp.symbols("x")
print(sp.solve(2*x - 10, x))
print(sp.solve(x**2 - 9, x))
print(sp.solve(sp.Eq(3*x, 12), x))
````

**Økonomisk tolkning:** Der inntekt minus kostnad er null, går bedriften i null (nullpunkt).

````python
q = sp.symbols("q")
inntekt = 120*q
kostnad = 4000 + 70*q
nullpunkt = sp.solve(inntekt - kostnad, q)
print(nullpunkt)
````

### Derivasjon og optimering
- Sett den deriverte lik null for å finne stasjonære punkter.
- Andrederivert > 0 gir minimum, < 0 gir maksimum.
- Sjekk at punktet ligger i det tillatte intervallet.

````python
q = sp.symbols("q")
kostnad = 2*q**2 - 24*q + 100
dk = sp.diff(kostnad, q)
punkt = sp.solve(dk, q)
d2k = sp.diff(kostnad, q, 2)
print("K'(q) =", dk)
print("stasjonært punkt:", punkt)
print("K''(q) =", d2k)
print("kostnad i punktet:", kostnad.subs(q, punkt[0]))
````

````python
qs = np.linspace(0, 12, 100)
K = sp.lambdify(q, kostnad)
plt.plot(qs, K(qs))
plt.axvline(6, linestyle="--")
plt.xlabel("Mengde q")
plt.ylabel("Kostnad")
plt.title("Kostnadsfunksjon med minimum")
plt.show()
````

**🧠 Forutsi først:** Er q = 6 et minimum eller maksimum, og hvorfor?

<details><summary>Vis svar</summary>

Minimum. Andrederivert er 4 > 0, parabelen vender opp, og K'(q) skifter fra negativ til positiv i q = 6.

</details>

    nå = 0
    rekord = 0
    for i in range(1, len(kurser)):
